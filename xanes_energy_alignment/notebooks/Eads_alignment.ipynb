```python

#energy alignment for (NH4)2MoS4
Efermi = 1.8166
ECH = -0.21365606E+03  # XCH
EGS = -0.28553067E+03 # SCF
ECORE = -19816.9363  # POTCAR


delta1 = (ECH - EGS) - (Efermi)  # Ecbm = Efermi. #delta 1= (Exch - EGS)-(Ecbm-Ecore)
print ("delta1:",delta1)


Eex_raw = 2505.19  # example


step1_shift = Eex_raw + delta1
# Print the result
print("step1_shift:", step1_shift)


Ref_value= 2522.4 # shift wrt L3 edge for (NH4)2MoS4 in ref paper: J. Phys. Chem. C 2021, 125, 32, 17761–17773

delta2= Ref_value - step1_shift
print("delta2:",delta2)

Eex_align= step1_shift + delta2
print("Eex_align:",Eex_align)

total_shift = delta1 + delta2
print("total shift:", total_shift)
```
